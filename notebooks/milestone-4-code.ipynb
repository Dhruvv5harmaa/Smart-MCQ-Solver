```python
import pandas as pd
import torch
import numpy as np

from transformers import (
    AutoTokenizer,
    AutoModelForMultipleChoice,
    Trainer,
    TrainingArguments
)

from datasets import Dataset

from peft import (
    LoraConfig,
    get_peft_model,
    TaskType
)

from torch.nn.functional import softmax
```

```python
train = pd.read_csv("/kaggle/input/competitions/smart-mcq-solver-challenge/train.csv")

train.head()
```

```python
label_map = {
    "A":0,
    "B":1,
    "C":2,
    "D":3,
    "E":4
}

train["label"] = train["answer"].map(label_map)

print("Q1 =", train.loc[150,"label"])
```

```python
MODEL = "bert-base-uncased"

tokenizer = AutoTokenizer.from_pretrained(MODEL)
```

```python
text = (
    str(train.loc[0,"prompt"])
    + " [SEP] "
    + str(train.loc[0,"B"])
)

print(text)

print("Q2 =", len(text))
```

```python
choices = []

for option in ["A","B","C","D","E"]:
    choices.append(
        str(train.loc[0,"prompt"])
        + " [SEP] "
        + str(train.loc[0,option])
    )

enc = tokenizer(
    choices,
    padding="max_length",
    truncation=True,
    max_length=128,
    return_tensors="pt"
)

input_ids = enc["input_ids"].unsqueeze(0)
attention_mask = enc["attention_mask"].unsqueeze(0)

print(input_ids.shape)

print("Q3 =", input_ids.shape[1])
```

```python
all_ids = []

for i in range(16):

    choices = []

    for option in ["A","B","C","D","E"]:
        choices.append(
            str(train.loc[i,"prompt"])
            + " [SEP] "
            + str(train.loc[i,option])
        )

    enc = tokenizer(
        choices,
        padding="max_length",
        truncation=True,
        max_length=128,
        return_tensors="pt"
    )

    all_ids.append(enc["input_ids"])

all_ids = torch.stack(all_ids)

print(all_ids.shape)

print("Q4 =", all_ids.numel())
```

```python
model = AutoModelForMultipleChoice.from_pretrained(MODEL)
```

```python
outputs = model(
    input_ids=input_ids,
    attention_mask=attention_mask
)

print(outputs.logits.shape)

print("Q5 =", outputs.logits.shape[1])
```

```python
label = torch.tensor([train.loc[0,"label"]])

outputs = model(
    input_ids=input_ids,
    attention_mask=attention_mask,
    labels=label
)

print(outputs.loss)

print(outputs.loss.shape)

print("Q6 =", outputs.loss.dim())
```

```python
config = LoraConfig(
    r=8,
    lora_alpha=16,
    target_modules=["query","value"],
    lora_dropout=0.1,
    bias="none",
    task_type=TaskType.SEQ_CLS
)

lora_model = get_peft_model(model, config)

trainable = sum(
    p.numel()
    for p in lora_model.parameters()
    if p.requires_grad
)

print("Q7 =", trainable)
```

```python
small = train.iloc[:100].copy()


def preprocess(example):

    texts = []

    for option in ["A","B","C","D","E"]:
        texts.append(
            str(example["prompt"])
            + " [SEP] "
            + str(example[option])
        )

    enc = tokenizer(
        texts,
        padding="max_length",
        truncation=True,
        max_length=128
    )

    enc["labels"] = example["label"]

    return enc


dataset = Dataset.from_pandas(small)

dataset = dataset.map(preprocess)

print(np.array(dataset[0]["input_ids"]).shape)

print("Q8 =", np.array(dataset[0]["input_ids"]).shape[0])
```

```python
train32 = train.iloc[:32].copy()

dataset = Dataset.from_pandas(train32)

dataset = dataset.map(preprocess)

dataset.set_format(
    type="torch",
    columns=[
        "input_ids",
        "attention_mask",
        "labels"
    ]
)

model = AutoModelForMultipleChoice.from_pretrained(MODEL)

config = LoraConfig(
    r=8,
    lora_alpha=16,
    target_modules=["query","value"],
    lora_dropout=0.1,
    bias="none",
    task_type=TaskType.SEQ_CLS
)

model = get_peft_model(model, config)

training_args = TrainingArguments(
    output_dir="mcq",
    max_steps=4,
    per_device_train_batch_size=4,
    gradient_accumulation_steps=1,
    learning_rate=2e-5,
    logging_steps=1,
    report_to="none"
)


trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=dataset
)

train_result = trainer.train()

print("Q9 =", trainer.state.global_step)

choices = []

for option in ["A","B","C","D","E"]:
    choices.append(
        str(train.loc[0,"prompt"])
        + " [SEP] "
        + str(train.loc[0,option])
    )

enc = tokenizer(
    choices,
    padding="max_length",
    truncation=True,
    max_length=64,
    return_tensors="pt"
)

with torch.no_grad():

    logits = model(
        input_ids=enc["input_ids"].unsqueeze(0),
        attention_mask=enc["attention_mask"].unsqueeze(0)
    ).logits

prob = softmax(logits, dim=1)

print(prob)

print("Q10 =", round(prob[0,4].item(),4))
```
